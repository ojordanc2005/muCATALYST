```python
from matplotlib import pyplot as plt
import numpy as np
import math
from scipy import special
#help(special)
```

```python
def Bessel_mth_order(x_vals, m):
    J = []
    for x in x_vals:
        l = 0
        sum = 0
        while l < 100:

            sum += (((-1)**l) / (math.factorial(l) * math.factorial(l + m))) * (x / 2)**((2 * l) + m)

            l += 1
        J.append(sum)
    return J

x_vals = np.linspace(0,20,100)

plt.plot(x_vals, Bessel_mth_order(x_vals, 0), color='red', label = '$J_0(x)$')
plt.plot(x_vals, Bessel_mth_order(x_vals, 1), color='green', label = '$J_1(x)$')
plt.plot(x_vals, Bessel_mth_order(x_vals, 2), color='blue', label = '$J_2(x)$')

plt.xlabel('$x$')
plt.legend()
plt.grid()
plt.show()
```

Scipy also has some fancy functions which should make this easier:

```python
# supposedly, Bessel functions of the first kind are computed with the special.jv(x) syntax

plt.plot(x_vals, special.jv(0, x_vals), color='red', label = '$J_0(x)$')
plt.plot(x_vals, special.jv(1, x_vals), color='green', label = '$J_1(x)$')
plt.plot(x_vals, special.jv(2, x_vals), color='blue', label = '$J_2(x)$')

plt.xlabel('$x$')
plt.legend()
plt.grid()
plt.show()
```
