```python
R_2 = 142
R_1 = 80
eta = 2.450 * R_2 / R_1

Cu_conduct = 5.8e7
Be_cunduct = 1.8e7

#surfact resistances
R_s2 = 1
R_s1 = 1

#hard coding relevant Bessel function values
J_1_0 = 0
J_1_eta = 0.087
J_0_eta = -0.394
J_2_eta = 0.442

#this value is however much the cavity is not cut-off
phase = 0.01
J_1_0 = J_1_0 + phase

A = R_s1 * ((R_1**2 * J_1_0**2) - (R_2**2 * J_1_eta**2) + (R_2**2 * J_0_eta * J_2_eta))
B = R_s2 * ((R_2**2 * J_1_eta**2) - (R_2**2 * J_0_eta * J_2_eta))

mu_0 = 1.3e-6
c = 3e8

L = np.linspace(0, .14, 20)
m = mu_0 * mu_0 * c * c / np.pi / (A + B)
R = m * L
```

```python
#i can find an approximate value for T^2 by fitting to a "curve"
y = [3.6e7, 3.8e7, 4.25e7, 4.4e7, 4.22e7, 3.85e7, 3.6e7]
x = [0.05, 0.06, 0.08, 0.10, 0.12, 0.14, 0.15]

plt.plot(x,y)
#plt.ylim(3.2e7, 4.6e7)

T = 132e-7

y_new = []
for L in y:
    y_new.append(L * c * c * mu_0 * mu_0 * T / (np.pi * (A + B)))

print(y)
plt.plot(x, y_new)

plt.show()
```

__The transitive time factor for a pillbox cavity with small velocity changes is indeed $$T = \frac{sin(\frac{\pi L}{\beta \lambda})}{\frac{\pi L}{\beta \lambda}}.$$__
