```python
phase_of_particles = distance_between_cavities / beta / c * 2 * np.pi * f
phase_of_wave = phase_offset / 180 * np.pi

phase_sum = phase_of_particles + phase_of_wave


print(f'phase of particles: {phase_of_particles}')
print(f'phase of wave: {phase_of_wave}')

print(f'phase sum: {phase_sum}')
print(f'phase sum goal: {2 * np.pi}')
```
