The sinc function is that of the transitive time factor:

```python
from matplotlib import pyplot as plt
import numpy as np

x = np.linspace(-5, 5, 100)

def sinc(x):
    y = np.sin(np.pi * x)/(np.pi * x)
    return y

y = sinc(x)
plt.plot(x,y)
plt.show()
```

```python
L = 142 # in mm
z = np.linspace(-L/2, L/2, 300)

beta = 0.9 # beta = v/c where v = 0.9c
wavelength = 8 # in ??

X = (2 * z) / (beta * wavelength)

y = sinc(X)
plt.plot(X,y)
plt.show()
```
