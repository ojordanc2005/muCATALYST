```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display
from scipy import special
```

```python
# MAKE SURE TO "RUN ALL" EVERY TIME YOU WANT ANIMATIONS!!! idk why :/
from processing_functions import *
# physical constants
c = 3e8
epsilon = 8.8e-12
mu = 1.3e-6
Z_0 = np.sqrt(mu / epsilon)

# cavity parameters
a = 0.14232 # in mm
L = 0.11334 # in mm
beta = 0.9
kappa = 58.0e6 # for copper, in S/m

f = mode_1_frequency(a)
period = 1/f
E_0 = max_field(a, L)
```

```python
#from matplotlib import rc
#rc('font',**{'family':'serif','serif':['Times']})
#rc('text', usetex=True)

nframes = 60
#plotcolor = 'slateblue'
plotcolor = '#4e63f8'
#csfont = {'fontname':'Comic Sans MS'}
#hfont = {'fontname':'Courier'}

#plt.title('title',**csfont)
#plt.xlabel('xlabel', **hfont)

#x_vals = np.linspace(-a -0.02, a+0.02, 20)
#y_vals = np.linspace(-a -0.02, a+0.02, 20)
x_vals = np.linspace(-a, a, 20)
y_vals = np.linspace(-a, a, 20)
t_vals = np.linspace(0,period,nframes)

X, Y = np.meshgrid(x_vals,y_vals)
U = 0
V = 0

fig, ax = plt.subplots(1,1)

Q = ax.quiver(X, Y, U, V, color=plotcolor, label='$E_{\z}$')
ax.set_ylabel('r')
#ax.set_xlabel('z', **hfont)
ax.set_title('Longitudinal oscillation of an E field in a pillbox cavity')

def update_quiver(frame_num, Q, X, Y):
    t = t_vals[frame_num]

    U = E_0 * special.jv(0, ((special.jn_zeros(0, 1) / a * Y))) * np.cos(t * 2 * np.pi * f)
    #U = E_field(E_0, a, Y, f, t)
    V = 0 * X
    
    #if Y > a:
    #    U = 0 * Y

    Q.set_UVC(U,V)

    return Q,

anim = FuncAnimation(fig, update_quiver, fargs=(Q, X, Y), frames = nframes, interval=50)
display(HTML(anim.to_jshtml()))

anim.save('animations/pillbox_E-field.mp4', writer='ffmpeg')
```

```python
nframes = 60

x_vals = np.linspace(-(a * np.sqrt(2)), (a * np.sqrt(2)), 20)
y_vals = np.linspace(-(a * np.sqrt(2)), (a * np.sqrt(2)), 20)
t_vals = np.linspace(0,period,nframes)

U = 0
V = 0

fig, ax = plt.subplots(1,1)

Q = ax.quiver(X, Y, U, V, color=plotcolor, label='$H_{\phi}$')
ax.set_ylabel('y')
ax.set_xlabel('x')
ax.set_title('Transverse oscillation of an H field in a pillbox cavity')

def update_quiver(frame_num, Q, X, Y):
    t = t_vals[frame_num]

    theta = np.arctan(Y / X)
    phi = (np.pi / 2) + theta
    radius = np.sqrt(X**2 + Y**2)

    #for some reason, using sin instead of cos here makes god himself spawn onto my animation to smite me... vv

    U = (E_0 / Z_0 * special.jv(1, ((special.jn_zeros(0, 1) / a * radius))) * np.cos(t * 2 * np.pi * f)) * np.cos(phi) * -X / abs(X)
    V = (E_0 / Z_0 * special.jv(1, ((special.jn_zeros(0, 1) / a * radius))) * np.cos(t * 2 * np.pi * f)) * np.sin(phi) * -X / abs(X)
    
    #if radius>a:
    #    U = 0 * X
    #    V = 0 * Y

    Q.set_UVC(U,V)

    return Q,

anim = FuncAnimation(fig, update_quiver, fargs=(Q, X, Y), frames = nframes, interval=50)
display(HTML(anim.to_jshtml()))

anim.save('animations/pillbox_H-field.mp4', writer='ffmpeg')
```
