```python
import numpy as np
from matplotlib import pyplot as plt
```

```python
# FOR THE LENGTH SWEEP OF THE CYLINDRICALLY-COUPLED WAVEGUIDE
Q_ex = np.array([5.562, 5.716, 5.817, 5.913, 6.078, 6.342, 6.632, 7.012, 7.423, 7.915, 8.441, 9.059, 9.758, 10.583, 11.452])
Q_tot = np.array([2.903, 3.246, 3.619, 4.014, 4.416, 4.886, 5.342, 5.887, 6.411, 7.013, 7.632, 8.336, 9.111, 10.008, 10.936])
radius = np.linspace(15, 40, 15) # in mm

Q_difference = Q_ex - Q_tot
plt.plot(radius, Q_difference)
plt.xlabel('Radius [mm]')
plt.ylabel('$Q_ex - Q_0$')

plt.title('Cylindrically coupled cell')
plt.show()
```

```python
# FOR THE LENGTH SWEEP OF THE RECTANGULARLY-COUPLED WAVEGUIDE
Q_ex = np.array([4.728, 4.712, 4.864, 5.216, 5.568, 6.341, 7.252, 8.759, 10.458, 12.742, 15.803, 20.731])
Q_tot = np.array([2.208, 2.384, 2.904, 3.362, 3.954, 5.050, 6.256, 8.018, 9.886, 12.298, 15.461, 20.477])
L = np.linspace(10, 100, 12) # in mm
Area = L*L

Q_difference = Q_ex - Q_tot
plt.plot(Area, Q_difference)
plt.xlabel('Area [$mm^2$]')
plt.ylabel('$Q_ex - Q_0$')

plt.title('Rectangularly coupled cell')
plt.show()
```

```python
# SINGLE-CELL COUPLING VS TWO-CELL COUPLING DIFFERENCE
# THE SINGLE-CELL SYSTEM WITH A HEIGHT OF 16mm AND RADIUS OF 40mm HAS Q_ex = 11.452 and Q_tot = 10.936.
# THE TWO-CELL SYSTEM WITH THE SAME PARAMETERS HAS Q_ex ~ 37 and Q_tot ~ 36.
```
