```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display
```

```python
# here's a 1D animation
def plotSine(y_vals, t_vals, nframes):
    marker1, = ax.plot([], marker= 'o', color = 'blue')

    def sineVals(frame_num):
        t = t_vals[frame_num]
        y = y_vals[frame_num]

        marker1.set_data([t],[y])
        return

    anim = FuncAnimation(fig, sineVals, frames=nframes, interval=20)
    display(HTML(anim.to_jshtml()))
    plt.close()

fig = plt.figure()
ax = plt.axes()
ax.grid()
ax.set_xlim(0, 2 * np.pi)
ax.set_ylim(-3, 3)

nframes = 60

period = np.pi * 2
omega = 2 * np.pi * (1/period)

t_vals = np.linspace(0,period,nframes)
y_vals = np.sin(t_vals * omega)

#plt.plot(t_vals, y_vals)

plotSine(y_vals, t_vals, nframes)
```

```python
# Source - https://stackoverflow.com/a/19338495
# Posted by ali_m, modified by community. See post 'Timeline' for change history
# Retrieved 2026-06-04, License - CC BY-SA 3.0


X, Y = np.mgrid[:2*np.pi:10j,:2*np.pi:5j]
U = np.cos(X)
V = np.sin(Y)

fig, ax = plt.subplots(1,1)
Q = ax.quiver(X, Y, U, V, pivot='mid', color='r', units='inches')

ax.set_xlim(-1, 7)
ax.set_ylim(-1, 7)

def update_quiver(num, Q, X, Y):
    """updates the horizontal and vertical vector components by a
    fixed increment on each frame
    """

    U = np.cos(X + num*0.1)
    V = np.sin(Y + num*0.1)

    Q.set_UVC(U,V)

    return Q,

# you need to set blit=False, or the first set of arrows never gets
# cleared on subsequent frames
nframes = 60
anim = FuncAnimation(fig, update_quiver, fargs=(Q, X, Y), frames = nframes, interval=50, blit=False)
display(HTML(anim.to_jshtml()))
```
